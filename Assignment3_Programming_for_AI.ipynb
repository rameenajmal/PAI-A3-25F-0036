## Question 3: Sensor cube (18 marks)
`data/sensor_cube.npy` holds temperature readings from 6 sensors over 10 days, 24 hours per day. Axes are (day, hour, sensor). `data/sensor_meta.npz` holds `offsets` (calibration per sensor), `sensor_ids` and `days`. The logger writes -999 when a reading fails. Some other readings are physically impossible for this site (below 0 or above 60 degrees C).

**3a (3 marks)** Load both files. Print the shape, dtype and size of the cube, list the arrays stored in the npz archive and print `sensor_ids`. Report how many readings equal -999 and how many other readings are impossible. Why is -999 more dangerous than NaN if you take a mean directly?

````python
import numpy as np

# load both files
cube = np.load('data/sensor_cube.npy')
meta = np.load('data/sensor_meta.npz')

print(cube.shape)
print(cube.dtype)
print(cube.size)

print(meta.files)
print(meta['sensor_ids'])

failed = (cube == -999)
print(failed.sum())

# other impossible readings: below 0 or above 60, but not -999
impossible = ((cube < 0) | (cube > 60)) & (cube != -999)
print(impossible.sum())
````

## Why is -999 more dangerous than NaN if you take a mean directly?


ans:
-999 is a real number, so np.mean includes it without any warning. One -999 pulls the mean far down and gives a wrong answer that looks normal. NaN is safer because the mean becomes NaN, so I can see that something is wrong. I can also use np.nanmean to skip the NaN values.

**3b (4 marks)** Make a float copy of the cube (not a view) where every failed or impossible reading is NaN. Prove with `np.shares_memory` that your copy is independent. Then apply calibration by adding `offsets` using broadcasting only, no loops. State the shape of the broadcast operand and why it aligns with the last axis. Write the check that the original file array is untouched.

````python

bad = (cube == -999) | (cube < 0) | (cube > 60)

# float copy of the cube, then set bad readings to NaN
clean = cube.astype(float)
clean[bad] = np.nan

# proof that clean is a separate copy
print(np.shares_memory(clean, cube))

offsets = meta['offsets']
print(offsets.shape)
calibrated = clean + offsets
print(calibrated.shape)

# check that the original array is untouched
original_again = np.load('data/sensor_cube.npy')
print(np.array_equal(cube, original_again))
print((cube == -999).sum())
````

**3c (4 marks)** Without loops compute: (i) the mean temperature of each sensor over all days and hours, (ii) the mean daily profile of each hour averaged over days and sensors (shape 24), (iii) the hottest (day, sensor) pair using daily means. NaN must not poison the result. State the axis you used for each and why.

````python
# mean of each sensor over all days and hours
sensor_mean = np.nanmean(calibrated, axis=(0, 1))
print(sensor_mean.shape)
print(np.round(sensor_mean, 2))

# mean daily profile: each hour averaged over days and sensors
hour_profile = np.nanmean(calibrated, axis=(0, 2))
print(hour_profile.shape)
print(np.round(hour_profile, 2))

# hottest (day, sensor) pair using daily means
daily_mean = np.nanmean(calibrated, axis=1)
print(daily_mean.shape)
flat_index = np.nanargmax(daily_mean)
day, sensor = np.unravel_index(flat_index, daily_mean.shape)
print(day, sensor, round(daily_mean[day, sensor], 2))

#I used np.nanmean so NaN does not ruin the result. 
# (i) axis=(0, 1) removes day and hour, leaving one mean per sensor, shape (6,). 
# (ii) axis=(0, 2) removes day and sensor, leaving one mean per hour, shape (24,). 
# (iii) axis=1 removes hour, giving daily means of shape (10, 6). 
# The hottest pair is day 9, sensor 4 (about 32.13 C).
````

**3d (4 marks)** For sensor T3 (index 2) compute the 3-hour moving average of the hourly series for day 4 (index 3) using `np.cumsum` and slicing only, not `np.convolve`. Fill NaN in that series with the mean of the series first. Report the length of the result and explain why it is not 24.

````python
# hourly series for sensor T3 on day 4
series = calibrated[3, :, 2].copy()
print(series.shape)

# filled NaN with the mean of the series
series_mean = np.nanmean(series)
series[np.isnan(series)] = series_mean

# 3-hour moving average with cumsum and slicing
w = 3
c = np.cumsum(series)
c = np.insert(c, 0, 0)
moving_avg = (c[w:] - c[:-w]) / w

print(len(moving_avg))
print(np.round(moving_avg, 2))

#took the 24 hourly values for sensor index 2 on day index 3 and filled NaN with the mean of the series. 
# Then I used np.cumsum with a 0 added at the start. The sum of any 3 hours is c[i+3] - c[i],
#  so (c[3:] - c[:-3]) / 3 gives the moving average. The result has length 22, not 24. A window needs 
# 3 values, so there are only 24 - 3 + 1 = 22 full windows. The first two hours do not have 3 values.
````
