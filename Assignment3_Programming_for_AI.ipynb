## Why is -999 more dangerous than NaN if you take a mean directly?

## Question 4: Weather stations with Pandas (20 marks)
`data/weather_stations.csv` contains daily records for five cities. The file is messy: mixed date formats (`YYYY-MM-DD` and `DD/MM/YYYY`), temperatures stored as text such as `34.2C`, -999 as failed rain gauge value, missing humidity, pressure readings of 0 and duplicate rows.

**4a (6 marks)** Load and clean. Print `info()` before and after. Remove exact duplicates, parse both date formats into one datetime column (a plain `pd.to_datetime` call without care will silently misread days and months, check that), convert temperature to float, turn -999 into NaN, turn pressure 0 into NaN, fill missing humidity with the median humidity of the same city. Report the number of rows removed and the number of values you changed in each column. Name the cleaned frame `weather`.

````python
import pandas as pd
import numpy as np

raw = pd.read_csv('data/weather_stations.csv')

print("BEFORE")
raw.info()

weather = raw.copy()

# step 1: remove duplicate rows
before = len(weather)
weather = weather.drop_duplicates()
weather = weather.reset_index(drop=True)
rows_removed = before - len(weather)

# step 2: dates (two formats, so parse each one separately)
date1 = pd.to_datetime(weather['date'], format='%Y-%m-%d', errors='coerce')
date2 = pd.to_datetime(weather['date'], format='%d/%m/%Y', errors='coerce')
weather['date'] = date1.fillna(date2)

# check: all dates should be between March and June 2026
print(weather['date'].isna().sum())
print(weather['date'].min())
print(weather['date'].max())

# step 3: temperature, remove C and make float
temp_text = weather['temp_c'].astype(str)
temp_changed = temp_text.str.endswith('C').sum()
weather['temp_c'] = temp_text.str.replace('C', '').astype(float)

# step 4: rain -999 becomes NaN
rain_changed = (weather['rain_mm'] == -999).sum()
weather['rain_mm'] = weather['rain_mm'].replace(-999, np.nan)

# step 5: pressure 0 becomes NaN
pressure_changed = (weather['pressure_hpa'] == 0).sum()
weather['pressure_hpa'] = weather['pressure_hpa'].replace(0, np.nan)

# step 6: missing humidity gets the median of the same city
humidity_changed = weather['humidity_pct'].isna().sum()
city_median = weather.groupby('city')['humidity_pct'].transform('median')
weather['humidity_pct'] = weather['humidity_pct'].fillna(city_median)

print("AFTER")
weather.info()

print("rows removed:", rows_removed)
print("temp_c changed:", temp_changed)
print("rain_mm changed:", rain_changed)
print("pressure_hpa changed:", pressure_changed)
print("humidity_pct changed:", humidity_changed)
````

**4b (5 marks)** Create a month column. Build a pivot table of mean temperature by city (rows) and month (columns), rounded to 1 decimal. Find the city with the largest rise from the first month to the last month. Add a column giving each row's temperature anomaly relative to its own city's mean, using `groupby` with `transform`.

````python
# step 1: month column
weather['month'] = weather['date'].dt.month

# step 2: pivot table, mean temperature by city and month
pivot = weather.pivot_table(values='temp_c', index='city', columns='month', aggfunc='mean')
print(pivot.round(1))

# step 3: rise from first month to last month
first_month = pivot.columns[0]
last_month = pivot.columns[-1]
rise = pivot[last_month] - pivot[first_month]
print(rise.round(1))
print(rise.idxmax())
print(round(rise.max(), 1))

# step 4: anomaly = temperature minus the mean of its own city
city_mean = weather.groupby('city')['temp_c'].transform('mean')
weather['temp_anomaly'] = weather['temp_c'] - city_mean
print(weather[['city', 'date', 'temp_c', 'temp_anomaly']].head())
````

**4c (5 marks)** A heat spell is 3 or more consecutive days with temp above 38 in one city. Find every spell (city, start date, length). You may use NumPy on each city's sorted series, for example `diff` on a boolean array, but explain your logic in comments. If your data contains a missing calendar day for a city, say what you did about it.

````python
spells = []

for city in weather['city'].unique():
    one = weather[weather['city'] == city]
    series = one.set_index('date')['temp_c'].sort_index()

    all_days = pd.date_range(series.index.min(), series.index.max())
    print(city, "missing days:", len(all_days) - len(series))
    series = series.reindex(all_days)

    # NaN > 38 is False, so a missing day breaks a spell
    hot = (series > 38).values

    # False at both ends so spells at the edges are found
    padded = np.concatenate([[False], hot, [False]])

    # diff: +1 = spell starts, -1 = spell ends
    change = np.diff(padded.astype(int))
    starts = np.where(change == 1)[0]
    ends = np.where(change == -1)[0]

    for i in range(len(starts)):
        length = ends[i] - starts[i]
        if length >= 3:
            spells.append([city, all_days[starts[i]].date(), length])

heat_spells = pd.DataFrame(spells, columns=['city', 'start_date', 'length'])
print(heat_spells)
````

**4d (4 marks)** Compute the correlation matrix of temperature, humidity, pressure and rain. Then compute it separately for Karachi only.
````python
cols = ['temp_c', 'humidity_pct', 'pressure_hpa', 'rain_mm']

# all cities together
corr_all = weather[cols].corr()
print(corr_all.round(2))

# Karachi only
karachi = weather[weather['city'] == 'Karachi']
corr_khi = karachi[cols].corr()
print(corr_khi.round(2))

# difference: Karachi minus all cities
print((corr_khi - corr_all).round(2))
````

Pooling cities can mislead because each city has a different average temperature and humidity, and mixing them hides the pattern inside a city. This is Simpson’s paradox.

Rain is mostly zero, so its correlations stay close to 0 (between -0.06 and 0.03). A correlation like this can hide a real link, because only a few rainy days carry any information.

## Question 5: Student performance (15 marks)
`data/student_performance.csv` has 240 students (some rows duplicated). Maximum marks are in the column names: quiz 20, assignment 20, midterm 25, final 35. Total = 100. Some midterms are the text `Absent`, some finals look like a typing slip (entered ten times too large), and some study hours are missing.

**5a (5 marks)** Clean the data. Absent counts as 0 for the total but must be kept in a separate boolean column. Fix finals that exceed 35 by the most defensible rule and say what rule you chose. Fill missing study hours with the section median. Create `total` and a `grade` column: A at 85 or more, B at 70 to 84.99, C at 55 to 69.99, D at 40 to 54.99, F below 40, using `np.where` (nested) or boolean masks, not a loop.

````python
import pandas as pd
import numpy as np

students = pd.read_csv('data/student_performance.csv')
print(len(students))

students = students.drop_duplicates().reset_index(drop=True)
print(len(students))

# keep Absent in its own column, then count it as 0
students['absent'] = students['midterm_25'] == 'Absent'
students['midterm_25'] = pd.to_numeric(students['midterm_25'], errors='coerce')
students['midterm_25'] = students['midterm_25'].fillna(0)

# finals above 35 are typing slips (10 times too big)
too_big = students['final_35'] > 35
print(too_big.sum())
students.loc[too_big, 'final_35'] = students['final_35'] / 10

section_median = students.groupby('section')['study_hours_week'].transform('median')
print(students['study_hours_week'].isna().sum())
students['study_hours_week'] = students['study_hours_week'].fillna(section_median)

students['total'] = (students['quiz_20'] + students['assignment_20']
                     + students['midterm_25'] + students['final_35'])

students['grade'] = np.where(students['total'] >= 85, 'A',
                    np.where(students['total'] >= 70, 'B',
                    np.where(students['total'] >= 55, 'C',
                    np.where(students['total'] >= 40, 'D', 'F'))))

print(students['grade'].value_counts())
print(students['total'].max())
````

**5b (5 marks)** Per section report: number of students, mean total, std of total, pass percentage (not F), and the student_id with the highest total. Then list students with attendance below 75 who still scored a B or better. Use a single boolean mask.

````python
students['passed'] = students['grade'] != 'F'

report = pd.DataFrame()
report['students'] = students.groupby('section')['total'].count()
report['mean_total'] = students.groupby('section')['total'].mean().round(2)
report['std_total'] = students.groupby('section')['total'].std().round(2)
report['pass_pct'] = (students.groupby('section')['passed'].mean() * 100).round(1)

top_rows = students.loc[students.groupby('section')['total'].idxmax()]
report['top_student'] = top_rows.set_index('section')['student_id']
print(report)

mask = (students['attendance_pct'] < 75) & students['grade'].isin(['A', 'B'])
low_attendance_good = students[mask]
print(len(low_attendance_good))
print(low_attendance_good[['student_id', 'section', 'attendance_pct', 'total', 'grade']])
````

**7a (4 marks)** Build the grid with `meshgrid` of 201 points per axis. Handle r = 0 correctly (the limit is 1; a plain division gives NaN, show you saw it and fixed it with `np.where`). Draw side by side: `pcolormesh` with a diverging colormap and symmetric `vmin`/`vmax`, and `contourf` with `contour` lines labelled using `clabel`. Add colorbars. Count (with Boolean sums) how many grid points have f below -0.1 and say what visual feature that corresponds to.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 201)
y = np.linspace(-10, 10, 201)
X, Y = np.meshgrid(x, y)
R = np.sqrt(X**2 + Y**2)

# plain division gives NaN at r = 0
f_plain = np.sin(R) / R
print(np.isnan(f_plain).sum())

# fix: use the limit 1 where r = 0
R_safe = np.where(R == 0, 1, R)
F = np.where(R == 0, 1, np.sin(R_safe) / R_safe)
print(np.isnan(F).sum())

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

pc = axes[0].pcolormesh(X, Y, F, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar(pc, ax=axes[0])
axes[0].set_title('pcolormesh')

cf = axes[1].contourf(X, Y, F, cmap='coolwarm', vmin=-1, vmax=1)
lines = axes[1].contour(X, Y, F, colors='black')
axes[1].clabel(lines)
plt.colorbar(cf, ax=axes[1])
axes[1].set_title('contourf with contour lines')

plt.show()

print((F < -0.1).sum())
````

**7c (3 marks)** Animate the travelling wave sin(r - 0.3 t)/r along the line y = 0 using `FuncAnimation` with 60 frames, `interval=50` and `blit=True`. Your update function must return the artist(s). Save as `wave.gif` with `PillowWriter`. Do not use `plt.show()` blocking calls. Say in a comment why blit needs the update function to return a tuple.

````python
import os
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter

x = np.linspace(-10, 10, 200)
r = np.abs(x)

def wave(t):
    return np.sin(r - 0.3 * t) / r

fig, ax = plt.subplots()
line, = ax.plot(x, wave(0))
ax.set_ylim(-2, 2)
ax.set_xlabel('x')
ax.set_ylabel('f')

def update(frame):
    line.set_ydata(wave(frame))
    # blit redraws only the returned artists, and it needs them as a tuple
    return (line,)

anim = FuncAnimation(fig, update, frames=60, interval=50, blit=True)

os.makedirs('outputs', exist_ok=True)
anim.save('outputs/wave.gif', writer=PillowWriter(fps=20))
plt.close()
````
